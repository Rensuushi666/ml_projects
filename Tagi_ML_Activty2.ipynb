```python
model_DecisionTreeClassifier = DecisionTreeClassifier(
    max_depth=3,
    random_state=42,
)
model_DecisionTreeClassifier.fit(x_train, y_train)
```

```python
features = [
    "age",
    "studytime",
    "failures",
    "absences",
    "G1",
    "G2",
]

x = df[features]
y= df ["pass"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
y_pred = model_DecisionTreeClassifier.predict(x_test)
print (y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Fail", "Pass"]
)

disp.plot(
)

plt.title("Student Pass/Fail Confusion matrix")
plt.show()
```

```python
results = x_test.copy()
results["Actual"] = y_test
results["Prediction"] = y_pred

results.head(10)
```

```python
results["Risk"]=results["Prediction"].map({
    0: "At Risk",
    1 : "Like to Pass"
})
results.head(10)
```

```python
results["Recommnedation"]=results["Prediction"].map({
    0:"Prioritize for academic intervention",
    1:"Continue regular monitoring"

})
results.head(10)
```
